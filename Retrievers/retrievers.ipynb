```python
from langchain_openai import OpenAIEmbeddings
from langchain_chroma import Chroma
from langchain_core.documents import Document
```

```python
documents = [
    Document(page_content="Virat Kohli is great at batting and he was the captain of Indian cricket team", metadata={"team": "India"}),
    Document(page_content="Rohit Sharma is great at batting and he is the current captain of Indian cricket team", metadata={"team": "India"}),
    Document(page_content="Jasprit Bumrah is great in balling and can bat as well", metadata={"team": "India"})
]
```

```python
vectorStore = Chroma.from_documents(documents=documents, embedding=OpenAIEmbeddings(), collection_name="sample", persist_directory="chroma_db")
```

```python
# convert vector store to a retriever
retriever = vectorStore.as_retriever(search_kwargs={"k": 2})
```

```python
query = "Who is the current captain of Indian cricket team?"
results = retriever.invoke(query)
```

```python
for i, doc in enumerate(results):
    print("result", i)
    print(doc.page_content)
```
