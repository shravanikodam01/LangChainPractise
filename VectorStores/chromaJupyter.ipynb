```python
from langchain_openai import OpenAIEmbeddings
from langchain_chroma import Chroma
from langchain_core.documents import Document
```

```python
doc1 = Document(page_content="Virat Kohli is great at batting and he was the captain of Indian cricket team", metadata={"team": "India"})

doc2 = Document(page_content="Rohit Sharma is great at batting and he is the current captain of Indian cricket team", metadata={"team": "India"})

docs = [doc1, doc2]
```

```python

vectorStore = Chroma(collection_name="sample", embedding_function=OpenAIEmbeddings(), persist_directory="chroma_db")
```

```python
vectorStore.add_documents(docs)
```

```python
# Semantic search
vectorStore.similarity_search(query="Who was the current captain of Indian cricket team in the past?", k=1)
```

```python
#Metadata filtering
vectorStore.similarity_search(query="", filter={"team": "Australia"})
```
